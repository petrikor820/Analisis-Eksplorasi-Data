```python
import numpy as np
import scipy.stats as stats

# 1. Generate 100 sample data points (e.g., from a normal distribution)
np.random.seed(42)  # For reproducibility
data = np.random.normal(loc=50, scale=10, size=100)

# ==========================================
# Bagian A: Estimasi Titik & Interval (Z-dist, alpha=0.05)
# ==========================================
mean_point_estimate = np.mean(data)
std_sample = np.std(data, ddof=1)
n = len(data)

# Z-score for 95% confidence level (alpha = 0.05)
z_score = stats.norm.ppf(0.975)
margin_of_error = z_score * (std_sample / np.sqrt(n))

ci_lower = mean_point_estimate - margin_of_error
ci_upper = mean_point_estimate + margin_of_error

print("--- BAGIAN A ---")
print(f"Estimasi Titik (Rata-rata): {mean_point_estimate:.4f}")
print(f"95% Confidence Interval (Z): [{ci_lower:.4f}, {ci_upper:.4f}]\n")

# ==========================================
# Bagian B: Resampling (Jackknife & Bootstrap)
# ==========================================

# --- Jackknife Resampling ---
jackknife_means = [np.mean(np.delete(data, i)) for i in range(n)]
jackknife_mean_estimate = np.mean(jackknife_means)

# --- Bootstrapping (1000 iterasi) ---
n_iterations = 1000
bootstrap_means = np.zeros(n_iterations)

for i in range(n_iterations):
    boot_sample = np.random.choice(data, size=n, replace=True)
    bootstrap_means[i] = np.mean(boot_sample)

bootstrap_mean_estimate = np.mean(bootstrap_means)

print("--- BAGIAN B ---")
print(f"Rata-rata Resampling Jackknife: {jackknife_mean_estimate:.4f}")
print(f"Rata-rata Resampling Bootstrap (1000x): {bootstrap_mean_estimate:.4f}")
```
